```python
import pandas as pd

df=pd.read_csv("../Telco-Customer-Churn.csv")
```

```python
print(df.shape)
print(df['TotalCharges'].dtype)
print(df['TotalCharges'].head(10))
```

```python
problem_rows=df[df['TotalCharges'].str.strip()=='']
print(f"Blank TotalCharges: {problem_rows.shape[0]}")
print(problem_rows[['tenure','MonthlyCharges','TotalCharges']])
```

```python
df['TotalCharges']=df['TotalCharges'].str.strip().replace('','0')
df['TotalCharges']=df['TotalCharges'].astype(float)
print(df['TotalCharges'].dtype)
print(f"NULL values:{df['TotalCharges'].isnull().sum()}")

```

```python
df=df.drop(columns=['customerID'])
print(df.shape)
```

```python
df['Churn']=df['Churn'].map({'Yes':1,'No':0})
print(df['Churn'].value_counts())
```

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
print(categorical_cols)
# one hot encoding for all categorical columns
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)
bool_cols = df.select_dtypes(include='bool').columns.tolist()
df[bool_cols] = df[bool_cols].astype(int)
print(df.shape)
print(df.head())
```

```python
print(df.isnull().sum().sum())
print(df.dtypes.value_counts())
```

```python
df.to_csv("../Telco-Customer-Churn-cleaned.csv", index=False)
print("Cleaned data saved to Telco-Customer-Churn-cleaned.csv")
print(df.shape)
```
